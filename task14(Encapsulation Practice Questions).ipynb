```python
#18. Class Name: Payment

from abc import ABC, abstractmethod

# Abstract class
class Payment(ABC):

    @abstractmethod
    def process_payment(self, amount):
        pass


# Subclass for Credit Card Payment
class CreditCardPayment(Payment):

    def process_payment(self, amount):
        print(f"Processing credit card payment of ₹{amount}")


# Subclass for UPI Payment
class UPIPayment(Payment):

    def process_payment(self, amount):
        print(f"Processing UPI payment of ₹{amount}")


# Example usage
credit = CreditCardPayment()
upi = UPIPayment()

credit.process_payment(5000)
upi.process_payment(1500)
```

```python
#24. Class Name: Employee

from abc import ABC, abstractmethod

class Employee(ABC):

    def __init__(self, name, salary):
        self.__name = name
        self.__salary = salary

    def get_name(self):
        return self.__name

    def get_salary(self):
        return self.__salary

    @abstractmethod
    def calculate_bonus(self):
        pass


class Manager(Employee):

    def __init__(self, name, salary, team_size):
        super().__init__(name, salary)
        self.team_size = team_size

    def calculate_bonus(self):
        return self.get_salary() * 0.20 + (self.team_size * 500)
        
        #__salary is private, it cannot be accessed directly in Manager, Developer, or HR.becouse these are child class 
         # self.get_salary() This calls the getter method and returns
                                                                 #bonus = 60000 * 0.20 + (8 * 500)
                                                                    #   = 12000 + 4000
                                                                   #   = 16000
class Developer(Employee):

    def __init__(self, name, salary, programming_language):
        super().__init__(name, salary)
        self.programming_language = programming_language

    def calculate_bonus(self):
        return self.get_salary() * 0.15


class HR(Employee):

    def __init__(self, name, salary, experience):
        super().__init__(name, salary)
        self.experience = experience

    def calculate_bonus(self):
        return self.get_salary() * 0.10 + (self.experience * 1000)


# Driver Code
m = Manager("sahil", 60000, 8)
d = Developer("soni", 40000, "Python")
h = HR("bhumi", 30000, 5)

print("Manager:", m.get_name())
print("Bonus:", m.calculate_bonus())

print("Developer:", d.get_name())
print("Language:", d.programming_language)
print("Bonus:", d.calculate_bonus())

print("\nHR:", h.get_name())
print("Experience:", h.experience)
print("Bonus:", h.calculate_bonus())
```

```python
#25. Class Name: Shape3D

from abc import ABC, abstractmethod
import math

class Shape3D(ABC):

    def __init__(self, name):
        self.__name = name

    def get_name(self):
        return self.__name

    @abstractmethod
    def volume(self):
        pass

    @abstractmethod
    def surface_area(self):
        pass


class Cube(Shape3D):

    def __init__(self,name, side):
        super().__init__(name)
        self.side = side

    def volume(self):
        return self.side ** 3

    def surface_area(self):
        return 6 * (self.side ** 2)

    def display (self):
        print("name is:",self.get_name())
        
class Sphere(Shape3D):

    def __init__(self,name, radius):
        super().__init__(name)
        self.radius = radius

    def volume(self):
        return (4/3) * math.pi * (self.radius ** 3)

    def surface_area(self):
        return 4 * math.pi * (self.radius ** 2)

    def display (self):
        print("name is:",self.get_name())

class Cylinder(Shape3D):

    def __init__(self, name,radius, height):
        super().__init__(name)
        self.radius = radius
        self.height = height

    def volume(self):
        return math.pi * (self.radius ** 2) * self.height

    def surface_area(self):
        return 2 * math.pi * self.radius * (self.radius + self.height)


    def display (self):
        print("name is:",self.get_name())


        
# Driver Code
cube = Cube("cube",3)
sphere = Sphere("sphere",3)
cylinder = Cylinder("cylinder",2, 10)


cube.display()
print("Volume:", cube.volume())
print("Surface Area:", cube.surface_area())


sphere.display()
print("Volume:", sphere.volume())
print("Surface Area:", sphere.surface_area())


cylinder.display()
print("Volume:", cylinder.volume())
print("Surface Area:", cylinder.surface_area())
```

```python
#26. Class Name: Notification

from abc import ABC, abstractmethod

class Notification(ABC):

    def __init__(self, sender):
        self.__sender = sender

    def get_sender(self):
        return self.__sender

    @abstractmethod
    def send(self, message):
        pass


class EmailNotification(Notification):

    def __init__(self, sender, email):
        super().__init__(sender)
        self.email = email

    def send(self, message):        
        print("From:", self.get_sender())
        print("Email:", self.email)
        print("Message:", message)
        
    def display(self):
        print("sender name is:",self.get_sender())

class SMSNotification(Notification):

    def __init__(self, sender, phone_number):
        super().__init__(sender)
        self.phone_number = phone_number

    def send(self, message):
        print("From:", self.get_sender())
        print("Phone:", self.phone_number)
        print("Message:", message)


    def display(self):
        print("sender name is:",self.get_sender())   


# Driver Code
email = EmailNotification("sahil", "sahil@567899.com")
sms = SMSNotification("bhumi", "+441234567890")



email.send("Your order has been shipped.")
email.display()
print()

sms.display()
sms.send("Your OTP is 123456.")
```
