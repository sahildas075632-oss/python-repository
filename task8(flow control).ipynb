```python
#15 Take three sides of a triangle. Check if they form a valid triangle. If 
#yes, also print if it is Equilateral, Isosceles, or Scalene.

a = float(input("Enter first side: "))
b = float(input("Enter second side: "))
c = float(input("Enter third side: "))
if (a+b>c) and (b+c>a) and (c+a>b) :
    print("valid triangle")
    if a==b==c:                               #problem
        print("equilateral")
    elif a==b and b==c and c==a:
        print("Isosceles")
    else:
        print("Scalene")
else:
    print("not valdid triangle")
```

```python
#16 Take student marks in 5 subjects. Calculate percentage and give
#grade: 90+ → A+, 80-89 → A, 70-79 → B, 60-69 → C, below 60 → Fail.

english=int(input("Enter the mark:"))
odia=int(input("Enter the mark:"))
hindi=int(input("Enter the mark:"))
science=int(input("Enter the mark:"))
math=int(input("Enter the mark:"))
totalmark=int(input("enter total mark:"))
obtenmark=english+odia+hindi+science+math
print(obtenmark)
percentage=obtenmark / totalmark*100
print(percentage)
if percentage>=90:
    print("grad A+")
elif percentage>80 and percentage<89:
    print("grad A")
elif percentage>70 and percentage<79:
    print("B")
elif percentage>60 and percentage<69:
    print("C")
elif percentage<60:
    print("Fail")
else:
    print("invalid mark")
```
