```python
#1. Create a list original = [1, 2, [3, 4], 5]. Make a shallow copy named shallow and change
#shallow[2][0] = 99. Print both lists and explain the result.

import copy
original = [1, 2, [3, 4], 5]

shallow = copy.copy(original)

print("shalow is:",shallow)

print("original is:",original)

shallow[2][0]=99

print("shalow is:",shallow)

print("original is:",original)

print("shalow is:",id(shallow))

print("original is:",id(original))

print("shalow is:",id(shallow[2][0]))

print("original is:",id(original[2][0]))

shallow[2].clear()                            #shallow[2][0].clear() not possible becouse int' object has no attribute 'clear'
                                              #shallow[0].clear() itt show 'int' object has no attribute 'clear'
print("shalow is:",shallow)                   #shallow[2].clear() it is supported becouse index 2 is a list

print("original is:",original)

shallow.clear()

print("shalow is:",shallow)

print("original is:",original)

```

```python
#2. Do the same as above but use deepcopy instead of shallow copy. Show the difference.

import copy

original = [1, 2, [3, 4], 5]

shallow = copy.deepcopy(original)

print("shalow is:",shallow)

print("original is:",original)

shallow[2][0]=99                 #it only chenge in sallow not in original

print("shalow is:",shallow)

print("original is:",original)


print("shalow is:",id(shallow))

print("original is:",id(original))

print("shalow is:",id(shallow[2][0]))

print("original is:",id(original[2][0]))

shallow[2].clear()                                 #hallow[2][0].clear() not possible becouse int' object has no attribute 'clear'
                                                   #shallow[0].clear() itt show 'int' object has no attribute 'clear'
print("shalow is:",shallow)                        #shallow[2].clear() it is supported becouse index 2 is a list

print("original is:",original)

shallow.clear()

print("shalow is:",shallow)

print("original is:",original)
```

```python
#5. Create a nested dictionary and perform shallow copy. Change a value in the inner dictionary and show the effect on original.

import copy
original = [1, 2, [3, 4], 5]
shallow = copy.copy(original)
print("shalow is:",shallow)
print("original is:",original)
shallow[2][0]=4
print("shalow is:",shallow)
print("original is:",original)
print("shalow is:",id(shallow[2][0]))
print("original is:",id(original[2][0]))
shallow[1]=4                               #Why did only shallow change?
                                           #Because shallow[1] is part of the outer list, and the outer list was copied separately. 
print("shalow is:",shallow)
print("original is:",original)
print("shalow is:",id(shallow[1]))
print("original is:",id(original[1]))
```

```python
import copy
original = [1, 2, [3, 4], 5]
shallow = copy.deepcopy(original)
print("shalow is:",shallow)
print("original is:",original)
shallow[2][0]=4
print("shalow is:",shallow)
print("original is:",original)
print("shalow is:",id(shallow[2][0]))
print("original is:",id(original[2][0]))
shallow[1]=4                                                                          
print("shalow is:",shallow)
print("original is:",original)
print("shalow is:",id(shallow[1]))
print("original is:",id(original[1]))
```

```python
#5. Bank Account
#Attributes
#• account_holder
#• balance
#Methods
#• deposit()
#• withdraw()
#Withdraw should happen only if balance is sufficient.

class Account:
    def __init__(self,account_holder,balance):
        self.Account_holder=account_holder
        self.Balances=balance
    def display(self):
        print(self.Account_holder)
        print(self.Balances)
    def deposit(self,amount):
        if amount>0:
            self.total_balance=self.Balances+amount
            print(f"total amount after depiosit:{self.total_balance}")
        else:
            print("enter a valid amount")
    def withdraw(self,amount):
        if amount>0:
            if amount<self.total_balance:
                rest_balance=self.total_balance-amount
                print(f"rest balance is:{rest_balance}")
            else:
                print("insofficient balance")
        else:
            print("plese enter a valid amount")
obj1=Account("sahil",23000)
obj1.display()
obj1.deposit(1000)
obj1.withdraw(2000)

        
```
