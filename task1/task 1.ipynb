```python
#simple calculater
X=int(input("Enter a:"))
Y=int(input("Enter b:"))
Z=input("enter operater :")
if Z=="+":
    print(X+Y)
elif Z=="-":
    print(X-Y)
elif Z=="*":
    print(X*Y)
elif Z=="/":
    print(X/Y)
elif Z=="//":
    print(X//Y)
elif Z=="**":
    print(X**Y)
elif Z=="%":
    print(X%Y)
else :
    print ("Error")
      
```
