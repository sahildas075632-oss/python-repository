```python
#16. Create a function calculate(a, b, operation) and show how keyword args improve readability.

def calculate(a, b, operation):
    if operation == "add":
        return a + b
    elif operation == "subtract":
        return a - b
    elif operation == "multiply":
        return a * b
    else:
        return "Invalid operation"
calculate(2,3,"subtract")
```

```python
#23. Write a function multiply(*args) that multiplies all numbers.

def multiply(*args):
    multiply=1
    for i in args:
        multiply=i*multiply
    return multiply
    
multiply(1,2,3,4)
```

```python
#24. Create a function that takes student marks as *args and returns average.

def average_mark(*args):
    t_sub=len(args)
    marks=sum(args)
    average=marks/t_sub
    return average
average_mark(50,50)
```
