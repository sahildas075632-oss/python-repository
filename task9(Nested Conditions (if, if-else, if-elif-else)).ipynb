```python
#11 Take a year as input. Check if it is a leap year using nested conditions.

year=input("Enter a year:")
if year.isdigit():
    year=int(year)
    if (year%4==0 and year%100!=0) or (year%400==0):
        print("leap year")
    else:
        print("not leap year")
else:
    print("invalid year")
```
