```python
#4 Remove element 20 from set s = {10, 20, 30, 40} using remove() and discard(). Show difference.

s={10,20,30,40}
s.remove(20)
print(s)                      #here diffsrence is remove method only remove the valu which is present in given set  
                              # it give error if the value is not in set                            
```

```python
s={10,20,30,40}
s.discard(50)
print(s)                       
s.discard(20)                
print(s)
                               #but discard method remove the valu also which is prasent in set but if the given value is not in
                               #set it return the original set in the output               
```

```python
#5  Create two sets A = {1, 2, 3, 4} and B = {3, 4, 5, 6}. Print union,intersection, and difference.

A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print(A.union(B))
print(A.intersection(B))
print(A.difference(B))
print(B.difference(A))
```

```python
#13 Take a character and check if it is a vowel (a, e, i, o, u) using if statement.

char=input("Enter a chracter:")
if char in ("a","e","i","O","U"):
    print("vowl")
else:
    print("consonant")

```
