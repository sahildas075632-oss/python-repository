```python
                #Bonus Mixed Question (String + Logical)

#Question 9:
#Python
# Take a password from user.
# Check if password is strong:
# - Length >= 8
# - Contains '@' or '#'
# Print "Strong Password" or "Weak Password"

password = input("Enter your password: ")

if len(password) >= 8 and ("@" in password or "#" in password):
    print("Strong Password")
else:
    print("Weak Password")
```

```python
                    #1. Logical Operators

#Python
# Take three subject marks from user (English, Math, Science).
# Check if the student has passed:
# All marks should be >= 40 (use 'and' operator)
# If any one subject is < 40, print "Fail"

english = int(input("Enter English marks: "))
math = int(input("Enter Math marks: "))
science = int(input("Enter Science marks: "))

if english >= 40 and math >= 40 and science >= 40:
    print("Pass")
else:
    print("Fail")
```

```python
                      #2. Logical Operators + User Input
#Python
# Take a number from user.
# Check whether the number is:
# Positive AND Even
# OR
# Negative OR Odd
# Print appropriate message

num = int(input("Enter a number: "))

if num >=0 and num % 2 == 0:
    print("Positive and Even")
else:
    print("Negative or Odd")
```

```python
                               #3. String Operations
#Python
# Take your city name as input.
# Print the following:
# 1. City name in reverse order
# 2. Number of vowels in the city name (a,e,i,o,u)
# 3. City name without first and last character

city = input("Enter your city name: ")

# 1. City name in reverse order
print("Reverse:", city[::-1])

# 2. Count number of vowels
count = 0

for ch in city.lower():
    if ch in "aeiou":
        count += 1

print("Number of vowels:", count)

# 3. City name without first and last character
print("Without first and last character:", city[1:6])
```

```python
                               #4. String Slicing
#Python
# Take a mobile number as input (10 digits).
# Print:
# First 3 digits (network code)
# Last 4 digits
# Middle 3 digits

mobile = input("Enter your 10-digit mobile number: ")
print("First 3 digits:", mobile[:3])
print("Last 4 digits:", mobile[6:])
print("Middle 3 digits:", mobile[3:6])
```

```python
                               #5. Mixed (String + Logical)
#Python
# Take a string from user.
# Check if:
# - The string length is greater than 5
# - AND it starts with 'A' or 'a'
# Print "Valid String" or "Invalid String"

text = input("Enter a string: ")

if len(text) > 5 and (text.startswith("A") or text.startswith("a")):
    print("Valid String")
else:
    print("Invalid String")
```
