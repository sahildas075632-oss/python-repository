                                 #Questions on List Comprehensions

```python
#1. Write a list comprehension to create a list of numbers from 1 to 10.

result=[i for i in range (1,100)]
print(result)
```

```python
#2. Create a list of squares of numbers from 1 to 10 using list comprehension.

result=[i**2 for i in range(1,11) ]
print(result)
```

```python
#3. Write a list comprehension to get even numbers from 1 to 20.

result=["Even"for i in range(1,20) if i%2==0]
print(result)
```

```python
#4. Convert the following loop into a list comprehension:
#result = []
#for i in range(10):
#result.append(i*3)

result=[i**3 for i in range(10)]
print(result)
```

```python
#5. What is the output of [x for x in range(5)]?

[x for x in range(5)]
```

```python
#6. Create a list of first 5 multiples of 7 using list comprehension.

num=[i*7 for i in range(1,6)]
print(num)
```

```python
#7. Write a list comprehension to convert all strings in a list to uppercase.

words = ["apple", "banana", "cat"]
upper=[i.upper() for i in  words]
print(upper)
```

```python
#8. Predict the output: [i**2 for i in range(1,6)]

[i**2 for i in range(1,6)]
```

```python
#9. Create a list containing numbers from 0 to 9 using list comprehension.

num=list([i for i in range(0,10)])
print(num)
```

```python
#10.Write a list comprehension to get only positive numbers from [-2, -1, 0, 1, 2,3].

num=[-2, -1, 0, 1, 2,3]
result=[f"{i}:positive no"for i in num if i>0]
print(result)
```
