```python
## 1

from abc import ABC,abstractmethod

class InsufficientBalanceError(Exception):
    pass

class InvalidAmountError(Exception):
    pass

class AccountNotFoundError(Exception):
    pass

class DailyLimitExceededError(Exception):
    pass

class Account(ABC):
    total_account=0    #Class Variable

#constructer
    def __init__(self,acc_no,name,balance):
       self.__Ac_no=acc_no
       self.__name=name
       self.__Bal=balance
       Account.total_account += 1 
#geeter
    def get_ac_no(self):
       return self.__Ac_no
    def get_name(self):
       return f"Name  is:{self.__name}"
    def get_bal(self):
       return  self.__Bal

#Setter
    def set_bal(self, balance):
        self.__Bal = balance        
    
#abstract method      
    @abstractmethod
    def deposit(self,amount):
        pass
    @abstractmethod
    def withdraw(self,amount):
        pass
    @abstractmethod
    def get_balance(self):
        pass
        
#Concrete method:   
    def show_account_info(self):
        print("wellcom to my Banak")
        print("account no is:",self.get_ac_no())
        print(self.get_name())
        print("balance is:",self.get_bal())
               
#classmethod(cls)    
    @classmethod 
    def show_total_accounts(cls):
        print("Total Account is:",cls.total_account)


#child class1
class SavingsAccount(Account):

    def __init__(self, acc_no, name, balance):
        super().__init__(acc_no,name,balance)
        self.__daily_withdrow = 0
        self.__interest_rate = 4
        
#geter
    def get_dely_withdrow(self):
        return self.__daily_withdrow

#geeter
    def get_interest(self):
        return self.__interest_rate
        
#getbalance
    def get_balance(self):
        return  self.get_bal()
        
#seeter
    def set_balance(self,amount):
        if amount>0:
            self.set_bal(amount)                  #here we call the perent class seter method and updet the new amount here
            print("new balance is:",self.get_bal())   #after updet the new amount  is display using get_bal() 
        else:
            raise InvalidAmountError("Balance cannot be negative")
            
#deposit            
    def deposit(self, amount):
        if amount > 0:                                          #if amount is 10 then and let self.get_bal is 5
            self.set_bal(self.get_bal() + amount)               #then (self.set_bal(5+10)) =  self.set_bal(15)
            print("Deposit Successful")
            print(" after deposit balance is:", self.get_bal())
        else:
            raise InvalidAmountError("Enter Valid Amount")
            
#withdrow
    def withdraw(self,amount):
        if amount<=0:
            raise InvalidAmountError("Amount must be greater than 0.")
            
        if self.__daily_withdrow + amount > 25000:
            raise DailyLimitExceededError("Daily withdrawal limit is 25000")

        if self.get_bal() - amount < 1000:
            raise InsufficientBalanceError("Minimum balance of 1000 must be maintained.")
     
           
        self.set_bal(self.get_bal() - amount)
        self.__daily_withdrow += amount
        print("Withdrow sussful remaning amount is:",self.get_bal())
        print(" withdrow amount is:",self.__daily_withdrow)
        
#intrest
    def calculate_interest(self):
        interest = (self.get_bal() * 4) / 100
        print("Intrest is:", interest ) 
        
    def show_account_info(self):
        print("wellcom to my Banak")
        print("account no is:",self.get_ac_no())
        print(self.get_name())
        print("balance is:",self.get_bal())
        print("dely withdraw lim is:",self.get_dely_withdrow())
        print("Interest is:",self.get_interest())

        
#child class2

class CurrentAccount(Account):

    def __init__(self, acc_no, name, balance):
        super().__init__(acc_no, name, balance)


        self.__overdraft_limit = 10000

# geter
    def get_overdraft_limit(self):
        return self.__overdraft_limit

# seter
    def get_balance(self):
        return self.get_bal()

# seter
    def set_balance(self, amount):
        if amount >= 0:
            self.set_bal(amount)
            print("new Balance is:", self.get_bal())
        else:
            raise InvalidAmountError("Balance cannot be negative.")

# deposit
    def deposit(self, amount):
        if amount >0:
            self.set_bal(self.get_bal() + amount)
            print("Deposit Successful")
            print(" after deposit balance is:", self.get_bal())
        else:
            raise InvalidAmountError("Amount must be greater than 0")

# withdrow
    def withdraw(self, amount):

        if amount <= 0:
            raise InvalidAmountError("Amount must be greater than 0")

        if self.get_bal() - amount < -self.__overdraft_limit:             #if self.get-bal()=30000,amount=1000
            raise InsufficientBalanceError("Overdraft limit of 10000 crosed")     #then 30000-1000< -10000
                                                                                    #2900<-10000
                                                                                    #here overdraft lim is not crosed so it allow
        self.set_bal(self.get_bal() - amount)
                                                                               #if it is like 1200<1000 then it not allow
        print("Withdrawal Successful Remaning Balance:", self.get_bal())
        

    def show_account_info(self):
        print("wellcom to my Banak")
        print("account no is:",self.get_ac_no())
        print(self.get_name())
        print("balance is:",self.get_bal())
        print("overdraped li is:",self.__overdraft_limit)
        print("No nterest") 




obj1= SavingsAccount(123,"sahil",2000)
obj2=CurrentAccount(234,"raj",3000)


obj1.deposit(1)
obj1.withdraw(200)

obj1.set_balance(3000)
obj1.calculate_interest()
obj1.show_account_info()
print("------------")
obj2.deposit(1000)
obj2.withdraw(200)

obj2.set_balance(50000)

obj2.show_account_info()
obj2.show_total_accounts()
```
