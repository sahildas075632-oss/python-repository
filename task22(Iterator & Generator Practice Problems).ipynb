```python
#Problem 6
#Create a custom iterator that prints numbers from 1 to 10.

class Number:
    def __init__(self):
        self.no=1
    def __iter__(self):
     return self
    def __next__(self):
        if self.no<=10:
            value=self.no
            self.no+=1
            return value
        else:
            raise StopIteration

obj=Number()
for i in obj:
    print(i)
```

```python
#8. Create a custom iterator that prints numbers in reverse order from 10 to 1.

class Number:
    def __init__(self):
        self.no=10
    def __iter__(self):
     return self
    def __next__(self):
        if self.no>=1:
            value=self.no
            self.no-=1
            return value
        else:
            raise StopIteration

obj=Number()
for i in obj:
    print(i)
```

```python
#9.Create a custom iterator that prints the characters of the word "GENERATOR" one by one.

class word:
    def __init__(self):
        self.char="GENERATOR"
        self.indent=0       
    def __iter__(self):
        return self
    def __next__(self):
         if self.indent<len(self.char):
             value=self.char[self.indent]
             self.indent+=1
             return value
         else:
            raise StopIteration

obj=word()
for i in obj:
    print(i)
```

```python
#10.Create a custom iterator that prints the multiplication table of 5. 


class Number:
    def __init__(self):
        self.no=1
    def __iter__(self):
     return self
    def __next__(self):
        if self.no<=10:
            value=self.no
            value*=5
            self.no+=1
            return value
        else:
            raise StopIteration

obj=Number()
for i in obj:
    print(i)
```
