```python
#3. Create Vehicle → Car, Electric → ElectricCar (ElectricCar inherits from both) Concept.


class Vehicle:

    def __init__(self, brand, **kwargs):
        super().__init__(**kwargs)
        self.__Brand = brand

    def get_brand(self):
        return self.__Brand

    def set_brand(self, new_brand):
        self.__Brand = new_brand
        print("New brand is:", self.__Brand)


class Car(Vehicle):

    def __init__(self, model, **kwargs):
        super().__init__(**kwargs)
        self.__Model = model

    def get_model(self):
        return self.__Model

    def set_model(self, new_model):
        self.__Model = new_model
        print("New model is:", self.__Model)


class Electric:

    def __init__(self, battery, **kwargs):#Since you are using super().__init__(**kwargs), pass the parent 
        super().__init__(**kwargs)        #class values through **kwargs.
        self.__Battery = battery

    def get_battery(self):
        return self.__Battery

    def set_battery(self, new_battery):
        self.__Battery = new_battery
        print("New battery is:", self.__Battery)


class ElectricCar(Car, Electric):

    def __init__(self, brand, model, battery, colour,**kwargs):
        super().__init__(brand=brand, model=model, battery=battery)
        self.__Colour = colour

    def get_colour(self):
        return self.__Colour

    def set_colour(self, new_colour):
        self.__Colour = new_colour
        print("New colour is:", self.__Colour)

    def display_all(self):
        print("Brand :", self.get_brand())
        print("Model :", self.get_model())
        print("Battery :", self.get_battery())
        print("Colour :", self.get_colour())


obj = ElectricCar("Tesla", "Model 3", "75 kWh", "White")

obj.display_all()

print("---------------------")

obj.set_brand("BMW")
obj.set_model("i4")
obj.set_battery("80 kWh")
obj.set_colour("Black")

print()

obj.display_all()
```

```python
#4. Demonstrate Hybrid Inheritance and print Method Resolution Order (MRO) Concept.


class A:

    def display(self):
        print("Display method from Class A")


class B(A):

    def display(self):
        print("Display method from Class B")


class C(A):

    def display(self):
        print("Display method from Class C")


class D(B, C):

    def display(self):
        print("Display method from Class D")


obj = D()

obj.display()

print(D.mro())
```

```python
#5. Create class A, B, C, D and solve the Diamond Problem using super() Concept.


class A:

    def __init__(self, **kwargs):
        super().__init__(**kwargs)
        print("Constructor of A")


class B(A):

    def __init__(self, **kwargs):
        super().__init__(**kwargs)
        print("Constructor of B")


class C(A):

    def __init__(self, **kwargs):
        super().__init__(**kwargs)
        print("Constructor of C")


class D(B, C):

    def __init__(self):
        super().__init__()
        print("Constructor of D")


obj = D()

print("\nMRO:", D.mro())
```
